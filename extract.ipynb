### Extracting

```python
import os
from moviepy.editor import *

# Tentukan path folder input dan output
input_folder = r'C:\Users\mefta\Desktop\VSC\Model Thesis\Thesis-Meftah\RealLifeDeceptionDetection.2016\Real-life_Deception_Detection_2016\Clips\Deceptive'
output_folder = r'C:\Users\mefta\Desktop\VSC\Model Thesis\Thesis-Meftah\RealLifeDeceptionDetection.2016\Real-life_Deception_Detection_2016\Audio\Deceptive'

# Pastikan folder output ada, jika belum, buat folder tersebut
if not os.path.exists(output_folder):
    os.makedirs(output_folder)

# List semua file dalam folder input
files = os.listdir(input_folder)

for file in files:
    if file.endswith(".mp4"):
        # Buat path lengkap dari file input dan output
        input_path = os.path.join(input_folder, file)
        output_path = os.path.join(output_folder, f"{os.path.splitext(file)[0]}.wav")

        # Load video dan ekstrak audio
        video = VideoFileClip(input_path)
        audio = video.audio

        # Simpan audio sebagai file WAV
        audio.write_audiofile(output_path, codec='pcm_s16le')

        # Tutup video setelah selesai diproses
        video.close()
```
